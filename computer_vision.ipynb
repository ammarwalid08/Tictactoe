```python
import cv2
flag = cv2.imread("flag.jpg")
print(flag)
```

```python
import matplotlib.pyplot as plt
plt.imshow(flag)
plt.show()

```

```python
flag = cv2.cvtColor(flag,cv2.COLOR_BGR2RGB)
plt.imshow(flag)
plt.show()
```

```python
img = cv2.cvtColor(flag,cv2.COLOR_RGB2GRAY)
plt.imshow(img,cmap="gray")
plt.show()
```

```python
crop_flag = flag[115:230,0:500]
plt.imshow(crop_flag)
plt.show()
```

```python
red, green, blue = cv2.split(flag)
plt.imshow(red,cmap = "Reds")
plt.show()
```
